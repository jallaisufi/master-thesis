```python
from pytreenet.special_ttn.binary import generate_binary_ttns
from pytreenet.operators.common_operators import ket_i
import numpy as np

phys_tensor = np.zeros((3, 2), dtype=complex)
phys_tensor[0, :] = ket_i(0, 2)
# Example for a 18 qubit circuit
quantum_circuit_state = generate_binary_ttns(
    num_phys=18, bond_dim=3, phys_tensor=phys_tensor, phys_prefix="qubit"
)

# print(quantum_circuit_state.tensors.keys())

for key in quantum_circuit_state.tensors.keys():
    print(key)
```
